# Cashiers Schema

Screenshot-ready display of the cashiers column types and shape.

```python
from pathlib import Path

import pandas as pd


columns = ["cashier_id", "shift", "flagged"]
candidate_paths = [
    Path("../../data/processed/Module 3/cashiers.csv"),
    Path("backend/Ai models/data/processed/Module 3/cashiers.csv"),
]

data_file = next((path for path in candidate_paths if path.exists()), None)
if data_file is None:
    raise FileNotFoundError("Could not find the Module 3 cashiers.csv file")

cashiers = pd.read_csv(data_file)[columns]

print("Column types:")
print(cashiers.dtypes)
print()
print(f"Shape: {cashiers.shape[0]} rows x {cashiers.shape[1]} columns")
```
