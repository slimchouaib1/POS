# Anomaly Detection Ground Truth Schema

Screenshot-ready display of the anomaly ground-truth column types and shape after Module 3 anomaly injection.

```python
from pathlib import Path

import pandas as pd


columns = ["order_id", "anomaly_type", "description"]
candidate_paths = [
    Path("../../data/processed/Module 3/anomalies_ground_truth.csv"),
    Path("backend/Ai models/data/processed/Module 3/anomalies_ground_truth.csv"),
]

data_file = next((path for path in candidate_paths if path.exists()), None)
if data_file is None:
    raise FileNotFoundError("Could not find the Module 3 anomalies_ground_truth.csv file")

anomalies = pd.read_csv(data_file)[columns]

print("Column types:")
print(anomalies.dtypes)
print()
print(f"Shape: {anomalies.shape[0]} rows x {anomalies.shape[1]} columns")
```
